## Task 6: Duplicate Checks
Check for duplicate records and duplicate employee IDs.

Your code should check:
- fully duplicated rows;
- duplicate `employee_id` values;
- the most common `employee_id` values using `value_counts()`.


````python
# Count of fully duplicated rows
full_duplicate_rows = df.duplicated().sum()
print(f"Number of fully duplicated rows: {full_duplicate_rows}")

#Count of duplicated employee id's
duplicated_employee_ids = df["employee_id"].duplicated().sum()
print(f"Number of employee ID duplicates: {duplicated_employee_ids}")

#Most Common Employee ID's (we shouldn't have any, but we do!)
print(df["employee_id"].value_counts().head(1))
````
